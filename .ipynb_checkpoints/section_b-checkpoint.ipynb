```python
from pyspark.sql import SparkSession
from operator import add

spark_session = SparkSession.builder \
    .master("spark://192.168.2.156:7077") \
    .appName("hao_bai_section_b") \
    .config("spark.dynamicAllocation.enabled", True) \
    .config("spark.dynamicAllocation.shuffleTracking.enabled", True) \
    .config("spark.shuffle.service.enabled", False) \
    .config("spark.dynamicAllocation.executorIdleTimeout", "30s") \
    .config("spark.cores.max", 4) \
    .getOrCreate()

sc = spark_session.sparkContext
```

```python
df = spark_session.read.csv("hdfs://192.168.2.156:9000/data/los-angeles-parking-citations/parking-citations.csv", header=True, inferSchema=True)

df.show(5)
```

```python
df.printSchema()
```

```python
print(f"Total number of rows: {df.count()}")
```

```python
print(f"Number of partitions: {df.rdd.getNumPartitions()}")
```

```python
df_cleaned = df.drop("VIN", "Latitude", "Longitude")
df_cleaned.show(5)
```

```python
from pyspark.sql.functions import col, max, count

df_cleaned = df_cleaned.withColumn("Fine amount", col("Fine amount").cast("float"))

max_fine = df_cleaned.select(max("Fine amount")).collect()[0][0]
print(f"Max fine amount: {max_fine}")

df_cleaned.filter(col("Fine amount") == max_fine).count()
```

```python
from pyspark.sql.functions import desc

df_cleaned.groupBy("Make") \
    .count() \
    .orderBy(desc("count")) \
    .show(20)
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

COLORS = {
    'AL': 'Aluminum', 'AM': 'Amber', 'BG': 'Beige', 'BK': 'Black',
    'BL': 'Blue', 'BN': 'Brown', 'BR': 'Brown', 'BZ': 'Bronze',
    'CH': 'Charcoal', 'DK': 'Dark', 'GD': 'Gold', 'GO': 'Gold',
    'GN': 'Green', 'GY': 'Gray', 'GT': 'Granite', 'IV': 'Ivory',
    'LT': 'Light', 'OL': 'Olive', 'OR': 'Orange', 'MR': 'Maroon',
    'PK': 'Pink', 'RD': 'Red', 'RE': 'Red', 'SI': 'Silver', 'SL': 'Silver',
    'SM': 'Smoke', 'TN': 'Tan', 'VT': 'Violet', 'WT': 'White', 'WH': 'White',
    'YL': 'Yellow', 'YE': 'Yellow', 'UN': 'Unknown'
}

def map_color(color):
    return COLORS.get(color, color)

map_color_udf = udf(map_color, StringType())

df_cleaned = df_cleaned.withColumn("color_long", map_color_udf(col("Color")))

df_cleaned.select("Color", "color_long").show(10)
```

```python
df_cleaned.filter(col("Make") == "TOYT") \
    .groupBy("color_long") \
    .count() \
    .orderBy(desc("count")) \
    .show(1)
```
