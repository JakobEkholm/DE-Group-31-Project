```python
from pyspark import SparkConf, SparkContext
from py4j.java_gateway import java_import

# 配置 Spark
conf = SparkConf().setAppName("DeleteHDFSFile").setMaster("local[*]")
sc = SparkContext(conf=conf)

# 确保 PySpark 访问 Hadoop FileSystem
java_import(sc._jvm, "org.apache.hadoop.fs.FileSystem")
java_import(sc._jvm, "org.apache.hadoop.fs.Path")

# 获取 HDFS 文件系统对象
fs = sc._jvm.org.apache.hadoop.fs.FileSystem.get(sc._jsc.hadoopConfiguration())

# 指定要删除的文件路径
file_path = "hdfs://192.168.2.156:9000/data/reddit/somefile.txt"

# 删除文件（True 表示递归删除，如果是目录也能删除）
fs.delete(sc._jvm.org.apache.hadoop.fs.Path(file_path), True)

print(f"File {file_path} deleted successfully.")
```
