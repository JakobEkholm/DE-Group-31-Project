```python
from pyspark.sql import SparkSession
from operator import add

spark_session = SparkSession.builder \
    .master("spark://192.168.2.156:7077") \
    .appName("hao_bai_section_a") \
    .config("spark.dynamicAllocation.enabled", True) \
    .config("spark.dynamicAllocation.shuffleTracking.enabled", True) \
    .config("spark.shuffle.service.enabled", False) \
    .config("spark.dynamicAllocation.executorIdleTimeout", "30s") \
    .config("spark.cores.max", 4) \
    .getOrCreate()

sc = spark_session.sparkContext
```

```python
print(sc.appName)
print(sc.master)
```

```python
en_rdd = sc.textFile("hdfs://192.168.2.156:9000/data/europarl/europarl-v7.sv-en.en")
sv_rdd = sc.textFile("hdfs://192.168.2.156:9000/data/europarl/europarl-v7.sv-en.sv")

en_lines = en_rdd.count()
sv_lines = sv_rdd.count()

print(f"English corpus lines: {en_lines}")
print(f"Swedish corpus lines: {sv_lines}")

assert en_lines == sv_lines, "Line counts do not match!"

print(f"English RDD partitions: {en_rdd.getNumPartitions()}")
print(f"Swedish RDD partitions: {sv_rdd.getNumPartitions()}")
```

```python
def preprocess(line):
    return line.lower().split()

en_rdd_clean = en_rdd.map(preprocess)
sv_rdd_clean = sv_rdd.map(preprocess)

print("English Sample:", en_rdd_clean.take(10))
print("Swedish Sample:", sv_rdd_clean.take(10))

assert en_rdd_clean.count() == sv_rdd_clean.count(), "Line counts do not match!"
```

```python
en_word_counts = en_rdd_clean.flatMap(lambda words: words) \
                             .map(lambda word: (word, 1)) \
                             .reduceByKey(lambda a, b: a + b) \
                             .sortBy(lambda x: x[1], ascending=False)

sv_word_counts = sv_rdd_clean.flatMap(lambda words: words) \
                             .map(lambda word: (word, 1)) \
                             .reduceByKey(lambda a, b: a + b) \
                             .sortBy(lambda x: x[1], ascending=False)

print("Top 10 English words:", en_word_counts.take(10))
print("Top 10 Swedish words:", sv_word_counts.take(10))
```

```python
en_indexed = en_rdd_clean.zipWithIndex().map(lambda x: (x[1], x[0]))
sv_indexed = sv_rdd_clean.zipWithIndex().map(lambda x: (x[1], x[0]))

aligned_rdd = en_indexed.join(sv_indexed)

filtered_rdd = aligned_rdd.filter(lambda x: len(x[1][0]) < 10 and len(x[1][1]) < 10)

filtered_rdd = filtered_rdd.filter(lambda x: len(x[1][0]) == len(x[1][1]))

word_pairs = filtered_rdd.flatMap(lambda x: zip(x[1][0], x[1][1]))

word_pair_counts = word_pairs.map(lambda pair: (pair, 1)) \
                             .reduceByKey(lambda a, b: a + b) \
                             .sortBy(lambda x: x[1], ascending=False)

print("Most common translations:", word_pair_counts.take(10))
```
